<span style='color: green;'>Let </span>$n_{\mathrm{obs}}$<span style='color: green;'> be the number of Clinton/Trump respondents used to calculate the raw estimate. Report </span>$n_{\mathrm{obs}}$<span style='color: green;'> and compute the conventional binomial standard error for </span>$\widehat{\mu}$<span style='color: green;'>. Explain in one sentence why this standard error does not include uncertainty from selection or nonresponse.</span>

```python
# Helped by LLMs: I asked ChatGPT how to calculate the
# conventional binomial standard error for a binary sample mean.

n_obs = len(two_party)

raw_se = (
    raw_estimate * (1 - raw_estimate) / n_obs
) ** 0.5

print("Number of respondents:", n_obs)
print("Standard error:", round(raw_se, 5))
```

This standard error only measures sampling variation among the 720 observed respondents and does not include uncertainty caused by who was selected into or chose to respond to the poll.

<span style='color: green;'>Draw </span>$n_{\mathrm{obs}}$<span style='color: green;'> observations with replacement from the Clinton/Trump respondents, calculate </span>$\widehat{\mu}$<span style='color: green;'>, and repeat this procedure 1,000 times. Plot a histogram of the resulting estimates and report their standard deviation.</span>

```python
# Helped by LLMs: I asked ChatGPT how to generate 1,000
# bootstrap samples, plot their estimates, and calculate the
# bootstrap standard deviation.

rng = np.random.default_rng(5355)
bootstrap_full = []

# Create 1,000 bootstrap samples
for i in range(1000):
    sample = rng.choice(
        two_party["clinton"],
        size=n_obs,
        replace=True
    )
    bootstrap_full.append(sample.mean())

bootstrap_full = np.array(bootstrap_full)

# Use these same settings for the next histogram
bins = np.linspace(0.45, 0.65, 31)

plt.hist(bootstrap_full, bins=bins)
plt.axvline(raw_estimate, color="black", linestyle="--")
plt.xlim(0.45, 0.65)
plt.xlabel("Bootstrapped Clinton Share")
plt.ylabel("Frequency")
plt.title("Bootstrap Distribution")
plt.show()

bootstrap_sd = bootstrap_full.std(ddof=1)

print("Bootstrap SD:", round(bootstrap_sd, 5))
print("Analytic SE:", round(raw_se, 5))
```

<span style='color: green;'>Repeat the previous procedure using </span>$n_{\mathrm{obs}}/2$<span style='color: green;'> observations in each bootstrap sample. Use the same bins and horizontal-axis limits for both histograms. In one sentence, describe how the spread changes.</span>

```python
# Helped by LLMs: I asked ChatGPT how to repeat the
# bootstrap with half the sample size while keeping the same
# histogram bins and horizontal-axis limits.

half_n = n_obs // 2
bootstrap_half = []

# Repeat the bootstrap using half the sample size
for i in range(1000):
    sample = rng.choice(
        two_party["clinton"],
        size=half_n,
        replace=True
    )
    bootstrap_half.append(sample.mean())

bootstrap_half = np.array(bootstrap_half)

# Use the same bins and limits as the first histogram
plt.hist(bootstrap_half, bins=bins)
plt.axvline(raw_estimate, color="black", linestyle="--")
plt.xlim(0.45, 0.65)
plt.xlabel("Bootstrapped Clinton Share")
plt.ylabel("Frequency")
plt.title("Bootstrap Distribution With Half the Sample")
plt.show()

print("Full-sample SD:", round(bootstrap_full.std(ddof=1), 5))
print("Half-sample SD:", round(bootstrap_half.std(ddof=1), 5))
```

The half-sample bootstrap is more spread out, with its standard deviation increasing from about 0.018 to about 0.026 because estimates become less stable when they are based on fewer observations.

<font color='green'> For each demographic type separately -- age, gender, party, race, and education -- weight the poll by just that demographic type, in accordance with the population proportions given. Report the resulting poll results, and briefly (at most 3 sentences) describe what you observe.

```python
# Helped by LLMs: I asked ChatGPT how to use group
# means and the supplied marginal proportions to calculate a
# separate weighted estimate for each demographic variable.

demographics = ["age", "gender", "party", "race", "education"]
estimates = {"Raw": raw_estimate}
results = []
marginal_estimates = {}

for variable in demographics:

    # Clinton proportion within each group
    group_means = (
        two_party.groupby(variable)["clinton"].mean()
    )

    # Target population proportions
    target_shares = (
        dfdemographic[
            (dfdemographic["Demographic_Type_1"] == variable)
            & (dfdemographic["Demographic_Type_2"].isna())
        ]
        .set_index("Demographic_1")["Electoral_Proportion"]
    )

    # Put the groups in the same order
    group_means = group_means.reindex(target_shares.index)

    # Calculate the weighted estimate
    clinton_estimate = np.average(
        group_means,
        weights=target_shares
    )

    marginal_estimates[variable] = clinton_estimate
    estimates["Marginal: " + variable] = clinton_estimate

    results.append({
        "Demographic": variable,
        "Clinton": clinton_estimate,
        "Trump": 1 - clinton_estimate
    })

marginal_results = pd.DataFrame(results)
display(marginal_results.round(3))
```

Party weighting causes the largest change, reducing Clinton’s estimate from 0.546 to 0.499 because the party mix of the poll differs from the target population. Age and race weighting also lower the estimate to about 0.531, while gender slightly raises it to 0.550. Education barely changes the result, giving an estimate of 0.544.

### 3) Two-dimensional joint-distribution weighting

<font color='green'>Now, for each pair of demographic types in dfdemographic, do the same -- weight the poll by that pair of demographic types, in accordance with the given joint distributions, and briefly (at most 3 sentences) describe what you observe.

For example, when weighted by race and age, you'll find:

Weighted by age and race: Clinton: 0.523, Trump: 0.477

```python
# Helped by LLMs: I asked ChatGPT how to loop through
# every supplied two-way demographic distribution and calculate the
# corresponding poststratified estimate.

# Find the ten supplied demographic pairs
pairs = (
    dfdemographic.dropna(subset=["Demographic_Type_2"])
    [["Demographic_Type_1", "Demographic_Type_2"]]
    .drop_duplicates()
)

results = []

for _, row in pairs.iterrows():

    variable_1 = row["Demographic_Type_1"]
    variable_2 = row["Demographic_Type_2"]
    pair_name = variable_1 + " + " + variable_2

    # Get the target proportions for this pair
    target = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == variable_1)
        & (dfdemographic["Demographic_Type_2"] == variable_2)
    ][
        ["Demographic_1", "Demographic_2", "Electoral_Proportion"]
    ].copy()

    target = target.rename(columns={
        "Demographic_1": variable_1,
        "Demographic_2": variable_2
    })

    # Find Clinton's mean within each pair of groups
    group_means = (
        two_party.groupby([variable_1, variable_2])["clinton"]
        .mean()
        .reset_index(name="group_mean")
    )

    # Match the poll means with the target proportions
    cells = target.merge(
        group_means,
        on=[variable_1, variable_2],
        how="left"
    )

    # Check whether any target group is missing from the poll
    if cells["group_mean"].isna().any():
        print("Missing poll cells for", pair_name)
        continue

    # Calculate the weighted result
    clinton_estimate = np.average(
        cells["group_mean"],
        weights=cells["Electoral_Proportion"]
    )

    estimates["Joint: " + pair_name] = clinton_estimate

    results.append({
        "Pair": pair_name,
        "Clinton": clinton_estimate,
        "Trump": 1 - clinton_estimate
    })

joint_results = pd.DataFrame(results)
display(joint_results.round(3))
```

The joint estimates range from about 0.494 to 0.548. Every pair that includes party produces an estimate close to 0.500, while most pairs without party remain more favorable to Clinton. This suggests that the incorrect party composition of the raw poll is a major source of the difference.

```python
# Helped by LLMs: I asked ChatGPT how to compare
# supplied joint proportions with products of marginal proportions
# for party-gender and race-gender.

# The file stores race-gender in the order gender-race
pairs = [
    ("party", "gender"),
    ("gender", "race")
]

weighted_results = []

for variable_1, variable_2 in pairs:

    pair_name = variable_1 + " + " + variable_2

    # Get the supplied joint proportions
    joint = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == variable_1)
        & (dfdemographic["Demographic_Type_2"] == variable_2)
    ][
        ["Demographic_1", "Demographic_2", "Electoral_Proportion"]
    ].copy()

    joint = joint.rename(columns={
        "Demographic_1": variable_1,
        "Demographic_2": variable_2,
        "Electoral_Proportion": "joint_share"
    })

    # Get the first variable's marginal proportions
    marginal_1 = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == variable_1)
        & (dfdemographic["Demographic_Type_2"].isna())
    ][["Demographic_1", "Electoral_Proportion"]].copy()

    marginal_1.columns = [variable_1, "share_1"]

    # Get the second variable's marginal proportions
    marginal_2 = dfdemographic[
        (dfdemographic["Demographic_Type_1"] == variable_2)
        & (dfdemographic["Demographic_Type_2"].isna())
    ][["Demographic_1", "Electoral_Proportion"]].copy()

    marginal_2.columns = [variable_2, "share_2"]

    # Multiply the marginal proportions
    product = marginal_1.merge(marginal_2, how="cross")
    product["independence_share"] = (
        product["share_1"] * product["share_2"]
    )

    # Combine the supplied and calculated proportions
    comparison = joint.merge(
        product,
        on=[variable_1, variable_2]
    )

    # Find Clinton's mean within each cell
    group_means = (
        two_party.groupby([variable_1, variable_2])["clinton"]
        .mean()
        .reset_index(name="clinton_mean")
    )

    comparison = comparison.merge(
        group_means,
        on=[variable_1, variable_2]
    )

    comparison["difference"] = (
        comparison["joint_share"]
        - comparison["independence_share"]
    )

    comparison["absolute_difference"] = (
        comparison["difference"].abs()
    )

    # Calculate both weighted estimates
    joint_estimate = np.average(
        comparison["clinton_mean"],
        weights=comparison["joint_share"]
    )

    independence_estimate = np.average(
        comparison["clinton_mean"],
        weights=comparison["independence_share"]
    )

    estimates["Independence: " + pair_name] = independence_estimate

    weighted_results.append({
        "Pair": pair_name,
        "Supplied Joint": joint_estimate,
        "Independence": independence_estimate
    })

    print(pair_name)
    display(comparison.round(3))

    print("Two largest cell differences:")
    display(
        comparison.sort_values(
            "absolute_difference",
            ascending=False
        ).head(2).round(3)
    )

weighted_results = pd.DataFrame(weighted_results)
display(weighted_results.round(4))
```

The supplied-joint and independence estimates are very close for both pairs: 0.5028 versus 0.5037 for party and gender, and 0.5344 versus 0.5351 for gender and race. The largest party-gender differences are for Democratic men and women, whose Clinton shares are 0.919 and 0.955, while the largest gender-race differences are for White men and women, whose Clinton shares are 0.327 and 0.459. These cells have different candidate preferences, so the independence assumption could matter even though the positive and negative share differences mostly offset in these two estimates.

### Bonus 2 (up to 1 point): Raking

<font color='green'> Hint: at the heart of raking is calculating a weight for each survey respondent, so that the weights, when summed up, match the population on desired marginals as much as possible. 

For example, suppose we have a survey with two respondents, A likes chocolate and B hates chocolate. We know that in the population, 80\% of people like chocolate (the number is made up). So in a representative sample with 2 respondents, we would expect to see 1.6 respondents who like chocolate. Thus, one possible thing to do is we assign A a weight of 1.6 and B a weight of 0.4, and then use these weights to weigh their answers to the question we care about.

Of course, when the number of respondents and the number of demographic dimensions that we want to match get larger, finding the weights itself becomes harder. There are mainly two ways to do this. The first is through an iterative approach known as Iterative Proportional Fitting (IPF). IPF iterates through each demographic variable at a time, and adjusts the weights of all respondents through post stratification. IPF is easier to implement by hand by just following the instructions. The second approach is known as generalized raking, which is done through solving an optimization problem, and proposed by [Deville, J. C., Särndal, C. E., & Sautory, O. (1993)](https://www.tandfonline.com/doi/abs/10.1080/01621459.1993.10476369). This approach attempts to solve one undesirable outcome of IPF that the resulting weights may be unbalanced, which leads to higher variance in the final estimate. [This blog post](https://dev.to/potloc/generalized-raking-for-survey-weighting-2d1d) gives a nice introduction to generalized raking, while providing a code snippet for implementing it in Python (you will have to make suitable changes, of course.)
